```python
area = "Isocortex"

trained_values = {}
untrained_values = {}

for i, data in enumerate(trained_data):
    v = data.loc[data["roi_name"] == area, "mean_lifetime"]
    animal = trained_mice[i]
    trained_values[animal] = np.nanmean(v)

for i, data in enumerate(untrained_data):
    v = data.loc[data["roi_name"] == area, "mean_lifetime"]
    animal = untrained_mice[i]
    untrained_values[animal] = np.nanmean(v)
```
