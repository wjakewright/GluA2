```python
def normalize_image(image, correction, slope):
    """Helper function to normalize images"""

    # Correct for baseline fluorescence
    corrected_image = image - correction
    normalized_image = corrected_image / slope

    return normalized_image
```

```python
PULSE_CORRECTION = 66.6
PULSE_SLOPE = 480.47
CHASE_CORRECTION = 67.7
CHASE_SLOPE = 1169.1

image_file = current_image

# Seperate the image channels
chase_image = image_file[:, :, 2]
pulse_image = image_file[:, :, 3]
```

```python
chase_norm = normalize_image(chase_image, CHASE_CORRECTION, CHASE_SLOPE)
pulse_norm = normalize_image(pulse_image, PULSE_CORRECTION, PULSE_SLOPE)
```

```python
import numpy as np
chase_norm[chase_norm < 0] = np.nan
pulse_norm[pulse_norm < 0] = np.nan
```

```python
plt.figure()
im1= plt.imshow(chase_norm[:, :])
plt.colorbar(im1, label='Data Value Units')
plt.figure()
im2 = plt.imshow(pulse_norm[:, :])
plt.colorbar(im2, label='Data Value Units')
```

```python
import numpy as np
esp = 1e-9
TIME_CONSTANT = 3
total_protein = chase_norm + pulse_norm
fraction_ratio = pulse_norm + total_protein
fraction_ratio = fraction_ratio + esp

# Calculate the lifetime
lifetime = np.absolute(TIME_CONSTANT / (np.log(1 / fraction_ratio)))
```

```python
plt.figure()
im3 = plt.imshow(lifetime, vmax=4)
plt.colorbar(im3, label='Data Value Units')
```

```python
from rasterio.features import rasterize
from shapely.geometry import shape, box, Polygon, MultiPolygon, mapping
def mask_from_geom(g, H, W):
    """Rasterize a single geometry to a boolean mask (all_touched=True)."""
    return rasterize(
        [(mapping(g), 1)], out_shape=(H, W), fill=0, dtype="uint8", all_touched=True
    ).astype(bool)
```

```python
def extract_roi_values(image_data, json_file):
    """"""
    masks = []
    H = image_data.shape[0]
    W = image_data.shape[1]
    # Extract all roi features
    features = [
        f
        for f in json_file.get("features", [])
        if isinstance(f, dict)
        and (f.get("geometry") or {}).get("type") in ("Polygon", "MultiPolygon")
    ]

    # Organize the rois and extract image data
    records = []
    id_to_feat = {}
    parent_to_children = defaultdict(list)
    id_to_name = {}
    # Iterate through each feature seperately
    for feature in features:
        properties = feature.get("properties") or {}
        measurements = properties.get("measurements") or {}
        orig_id = measurements.get("ID")
        parent_id = measurements.get("Parent ID")
        # Parse altas area name and hemisphere
        area, side = qutils.parse_area_side(properties)
        ## Get base area name and determine if it is a layer
        name = qutils.normalize_name(area)
        # Store name and feature relationships
        if orig_id is not None:
            id_to_feat[orig_id] = feature
            id_to_name[orig_id] = name
        if parent_id is not None:
            parent_to_children[parent_id].append(orig_id)
        # fix clip of the roi geometry
        geometry = qutils.fix_clip(feature["geometry"], H=H, W=W)
        # Extract the pixel lifetimes
        ## Move on if roi is empty
        if geometry is None:
            record_dict = {
                "level": None,
                "roi_name": name,
                "side": side,
                "parent_name": None,
                "orig_id": orig_id,
                "parent_id": parent_id,
                "n_pixels": 0,
                "mean_pulse": 0.0,
                "mean_chase": 0.0,
                "mean_lifetime": np.nan,
            }
            records.append(record_dict)
            continue

        ## Get mask
        mask = mask_from_geom(geometry, H, W)

        masks.append(mask)

    return masks
```

```python
from collections import defaultdict
import qupath_utils as qutils

masks = extract_roi_values(lifetime, current_atlas)
```
